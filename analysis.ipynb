```python
df = pd.read_csv("past_race.csv")
s = df["popularity"]
winrate = {1:0.30,2:0.19,3:0.14,4:0.11,5:0.08,6:0.06,7:0.04,8:0.03,9:0.02}
df["ex_win"] = (s.map(winrate))
df["is_win"] = (df["result"] == 1)
df["payout"] = (df["is_win"] * df["odds"])
print(df["payout"].mean())
df.groupby("track")[["ex_win","is_win","payout"]].mean()
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("keiba.db")
races = pd.read_csv("past_race.csv")
races.to_sql("races", conn, if_exists="replace", index=False)

def q(sql):
    return pd.read_sql_query(sql, conn)
```

```python
q("SELECT * FROM races")
```

```python
q("SELECT date,race_number,horse_name,result FROM races WHERE track = '芝'")
```

```python
q("SELECT AVG(result) AS avg_result FROM races")
```

```python
q("SELECT AVG(result = 1) AS avg_winrate FROM races")
```

```python
q("""
SELECT
     track,AVG(result = 1) AS avg_winrate,
     AVG(result <= 3) AS avg_placebet,
     COUNT(*) AS race_count
FROM races
GROUP BY track""")
```

```python
q("SELECT AVG((result = 1) * odds) AS return_rate FROM races")
```

```python
pop_rates = pd.DataFrame({"popularity": [1,2,3,4,5,6,7,8,9],"win_rate": [0.30,0.19,0.14,0.11,0.08,0.06,0.04,0.03,0.02]})
pop_rates.to_sql("pop_rates", conn, if_exists="replace", index=False)
```

```python
q("SELECT * FROM pop_rates")
```

```python
q("SELECT COUNT(*) FROM races")
```

```python
q("""
SELECT *
FROM races
JOIN pop_rates ON races.popularity = pop_rates.popularity
""")
```

```python
q("""
SELECT AVG(win_rate) AS ex_win,COUNT(*)
FROM races
JOIN pop_rates ON races.popularity = pop_rates.popularity
""")
```
